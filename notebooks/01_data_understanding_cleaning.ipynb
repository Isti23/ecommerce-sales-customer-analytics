##Overview

Tahap ini bertujuan mengidentifikasi masalah kualitas data yang dapat memengaruhi hasil analisis penjualan dan pelanggan.

##Missing Values Findings

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage (%)": (df.isnull().sum() / len(df) * 100).round(2)
})

display(missing_summary)
```

##Duplicate Records Assessment

```python
print("Jumlah baris duplikat:", df.duplicated().sum())
```

```python
duplicates = df[df.duplicated(keep=False)]

display(duplicates.head(10))
```

##Cancellation Transactions Assessment

```python
cancellation_mask = df["InvoiceNo"].str.startswith("C", na=False)

print("Jumlah baris pembatalan:", cancellation_mask.sum())

print(
    "Persentase baris pembatalan:",
    round(cancellation_mask.mean() * 100, 2),
    "%"
)
```

Hasil pemeriksaan menunjukkan bahwa terdapat 9.288 baris dengan InvoiceNo yang diawali huruf C, yaitu sekitar 1,71% dari total baris dataset.

##Negative Quantity Findings

```python
negative_qty_mask = df["Quantity"] < 0

print("Jumlah baris dengan Quantity negatif:", negative_qty_mask.sum())

print(
    "Persentase Quantity negatif:",
    round(negative_qty_mask.mean() * 100, 2),
    "%"
)
```

##Unit Price Assessment

```python
zero_price_mask = df["UnitPrice"] == 0
negative_price_mask = df["UnitPrice"] < 0

print("Baris dengan UnitPrice = 0:", zero_price_mask.sum())
print("Baris dengan UnitPrice < 0:", negative_price_mask.sum())
```

```python
display(df[df["UnitPrice"] < 0])
```

Hasil pemeriksaan menunjukkan bahwa terdapat 2.515 baris dengan UnitPrice bernilai nol dan 2 baris dengan UnitPrice negatif.

Pemeriksaan terhadap dua baris dengan UnitPrice negatif menunjukkan bahwa keduanya memiliki StockCode B dan Description berupa "Adjust bad debt". Kedua baris juga tidak memiliki CustomerID.

Berdasarkan deskripsinya, catatan tersebut tampaknya berkaitan dengan penyesuaian piutang tak tertagih, bukan penjualan produk retail biasa. Oleh karena itu, kedua baris perlu dipisahkan dari analisis penjualan produk apabila tidak sesuai dengan definisi revenue yang digunakan.

##Zero Quantity Assessment

```python
zero_qty_mask = df["Quantity"] == 0

print("Baris dengan Quantity = 0:", zero_qty_mask.sum())
```

```python
data_type_summary = pd.DataFrame({
    "Data Type": df.dtypes.astype(str),
    "Unique Values": df.nunique(),
    "Missing Values": df.isnull().sum()
})

display(data_type_summary)
```

Hasil pemeriksaan menunjukkan bahwa kolom InvoiceDate sudah menggunakan tipe data datetime64[ns], sehingga dapat digunakan untuk analisis berdasarkan tanggal dan waktu. Kolom Quantity menggunakan tipe data int64, sedangkan UnitPrice menggunakan float64, yang sesuai untuk perhitungan kuantitas dan harga.

Kolom InvoiceNo, StockCode, Description, dan Country menggunakan tipe data object. Tipe data tersebut sesuai untuk menyimpan nomor transaksi, kode produk, deskripsi produk, dan nama negara.

Sementara itu, kolom CustomerID menggunakan tipe data float64. Karena kolom ini berfungsi sebagai identitas pelanggan, tipe data tersebut perlu ditinjau kembali pada tahap Data Cleaning. Selain itu, kolom CustomerID dan Description masih memiliki nilai kosong berdasarkan hasil pemeriksaan sebelumnya.

```python
print("InvoiceNo berawalan C dan Quantity negatif:",
      (cancellation_mask & negative_qty_mask).sum())

print("InvoiceNo berawalan C tetapi Quantity tidak negatif:",
      (cancellation_mask & ~negative_qty_mask).sum())

print("Quantity negatif tetapi InvoiceNo tidak berawalan C:",
      (~cancellation_mask & negative_qty_mask).sum())
```

Hasil pemeriksaan menunjukkan bahwa terdapat 9.288 baris yang memiliki InvoiceNo berawalan huruf C sekaligus memiliki Quantity negatif. Seluruh baris dengan InvoiceNo berawalan C termasuk dalam kelompok tersebut, karena tidak ditemukan baris dengan InvoiceNo berawalan C yang memiliki Quantity nol atau positif.

Selain itu, ditemukan 1.336 baris dengan Quantity negatif yang tidak memiliki InvoiceNo berawalan C. Temuan ini menunjukkan bahwa kuantitas negatif tidak selalu ditandai dengan awalan C pada nomor invoice.

##Extreme Values Assessment

```python
print("5 nilai Quantity terkecil:")
display(df["Quantity"].nsmallest(5))

print("5 nilai Quantity terbesar:")
display(df["Quantity"].nlargest(5))

print("5 nilai UnitPrice terkecil:")
display(df["UnitPrice"].nsmallest(5))

print("5 nilai UnitPrice terbesar:")
display(df["UnitPrice"].nlargest(5))
```

```python
extreme_transactions = df.loc[
    df["Quantity"].isin([-80995, 80995]) |
    (df["UnitPrice"] == df["UnitPrice"].max())
]

display(
    extreme_transactions[
        ["InvoiceNo", "StockCode", "Description", "Quantity",
         "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
    ]
)
```

Pemeriksaan lebih lanjut menunjukkan bahwa nilai ekstrem memiliki konteks transaksi yang berbeda. Transaksi dengan InvoiceNo C556445 memiliki StockCode M, Description "Manual", Quantity -1, dan UnitPrice 38.970,00. Catatan ini perlu diperlakukan secara khusus karena tidak menggambarkan penjualan produk retail biasa.

Selain itu, ditemukan transaksi penjualan sebanyak 80.995 unit produk PAPER CRAFT, LITTLE BIRDIE pada invoice 581483, yang diikuti transaksi dengan kuantitas -80.995 pada invoice C581484. Pola tersebut mengindikasikan adanya pembatalan atas transaksi dengan jumlah besar. Oleh karena itu, nilai ekstrem perlu dipertimbangkan berdasarkan konteksnya, bukan langsung dihapus hanya karena nilainya jauh dari nilai umum.

# 3. Data Cleaning & Preparation

Tahap Data Cleaning & Preparation dilakukan untuk menangani masalah kualitas data yang telah ditemukan pada tahap sebelumnya. Proses ini mencakup penanganan nilai kosong, duplikasi, transaksi pembatalan, nilai yang tidak sesuai untuk analisis penjualan, serta penyesuaian tipe data jika diperlukan.

Seluruh proses pembersihan dilakukan pada salinan dataset agar data asli tetap terjaga dan perubahan yang dilakukan dapat ditelusuri.

```python
df_clean = df.copy()
```

##3.1 Penanganan Duplikasi



```python
duplicate_count_before = df_clean.duplicated().sum()

df_clean = df_clean.drop_duplicates()

print("Duplikat sebelum cleaning:", duplicate_count_before)
print("Jumlah baris setelah cleaning:", len(df_clean))
print("Duplikat tersisa:", df_clean.duplicated().sum())
```

**Hasil Penanganan Duplikasi**

Proses cleaning berhasil menghapus 5.268 baris duplikat dan menghasilkan 536.641 baris data. Setelah penghapusan, tidak ditemukan baris duplikat berdasarkan pemeriksaan seluruh kolom. Proses ini dilakukan pada salinan dataset sehingga data asli tetap tidak berubah.

##3.2 Penanganan Transaksi Pembatalan

```python
cancellation_mask_clean = df_clean["InvoiceNo"].str.startswith("C", na=False)

df_cancellations = df_clean[cancellation_mask_clean].copy()
df_clean = df_clean[~cancellation_mask_clean].copy()

print("Jumlah transaksi pembatalan yang dipisahkan:", len(df_cancellations))
print("Jumlah baris dataset utama:", len(df_clean))
print("Jumlah baris dataset pembatalan:", len(df_cancellations))
```

##3.3 Pemeriksaan Quantity Negatif pada Dataset Utama

```python
negative_qty_remaining = df_clean[df_clean["Quantity"] < 0]

print("Jumlah baris Quantity negatif yang tersisa:", len(negative_qty_remaining))

display(
    negative_qty_remaining[
        ["InvoiceNo", "StockCode", "Description", "Quantity",
         "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
    ].head(10)
)
```

```python
negative_qty_remaining = df_clean[df_clean["Quantity"] < 0]

print("Description kosong:",
      negative_qty_remaining["Description"].isna().sum())

print("UnitPrice = 0:",
      (negative_qty_remaining["UnitPrice"] == 0).sum())

print("CustomerID kosong:",
      negative_qty_remaining["CustomerID"].isna().sum())

print("Ketiganya kosong/nol secara bersamaan:",
      (
          negative_qty_remaining["Description"].isna()
          & (negative_qty_remaining["UnitPrice"] == 0)
          & negative_qty_remaining["CustomerID"].isna()
      ).sum())
```

```python
negative_adjustments_mask = df_clean["Quantity"] < 0

df_negative_adjustments = df_clean[negative_adjustments_mask].copy()
df_clean = df_clean[~negative_adjustments_mask].copy()

print("Baris Quantity negatif yang dipisahkan:",
      len(df_negative_adjustments))
print("Jumlah baris dataset utama:", len(df_clean))
print("Jumlah baris dataset penyesuaian:",
      len(df_negative_adjustments))
print("Quantity negatif tersisa di dataset utama:",
      (df_clean["Quantity"] < 0).sum())
```

**Hasil Penanganan Quantity Negatif**

Sebanyak 1.336 baris dengan Quantity negatif yang tidak memiliki awalan C pada InvoiceNo dipisahkan ke dalam dataset `df_negative_adjustments`. Setelah pemisahan, dataset utama berisi 526.054 baris dan tidak memiliki Quantity negatif. Baris yang dipisahkan tetap disimpan agar dapat ditinjau kembali apabila diperlukan.

##3.4 Pemeriksaan dan Penanganan UnitPrice

```python
print("Baris UnitPrice = 0:", (df_clean["UnitPrice"] == 0).sum())
print("Baris UnitPrice < 0:", (df_clean["UnitPrice"] < 0).sum())
print("Jumlah baris dataset utama:", len(df_clean))
```

```python
zero_price_rows = df_clean[df_clean["UnitPrice"] == 0]

print("Jumlah baris UnitPrice = 0:", len(zero_price_rows))
print("Quantity positif:", (zero_price_rows["Quantity"] > 0).sum())
print("Quantity negatif:", (zero_price_rows["Quantity"] < 0).sum())

display(
    zero_price_rows[
        ["InvoiceNo", "StockCode", "Description", "Quantity",
         "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
    ].head(10)
)
```

```python
zero_price_mask = df_clean["UnitPrice"] == 0

df_zero_price = df_clean[zero_price_mask].copy()
df_clean = df_clean[~zero_price_mask].copy()

print("Baris harga nol yang dipisahkan:", len(df_zero_price))
print("Jumlah baris dataset utama:", len(df_clean))
print("Baris UnitPrice = 0 yang tersisa:", (df_clean["UnitPrice"] == 0).sum())
```

Hasil penanganan menunjukkan bahwa 1.174 baris dengan UnitPrice nol telah dipisahkan ke dalam dataset `df_zero_price`. Setelah proses ini, dataset utama berisi 524.880 baris dan tidak memiliki UnitPrice nol. Baris yang dipisahkan tetap disimpan untuk menjaga keterlacakan data dan memungkinkan pemeriksaan lebih lanjut.

```python
negative_price_mask = df_clean["UnitPrice"] < 0

df_negative_price = df_clean[negative_price_mask].copy()
df_clean = df_clean[~negative_price_mask].copy()

print("Baris harga negatif yang dipisahkan:", len(df_negative_price))
print("Jumlah baris dataset utama:", len(df_clean))
print("Baris UnitPrice negatif yang tersisa:",
      (df_clean["UnitPrice"] < 0).sum())
```

Dua baris dengan UnitPrice negatif berhasil dipisahkan ke dalam dataset `df_negative_price`. Berdasarkan pemeriksaan sebelumnya, kedua baris tersebut memiliki deskripsi "Adjust bad debt" sehingga tampaknya berkaitan dengan penyesuaian piutang, bukan penjualan produk retail biasa. Setelah pemisahan, dataset utama berisi 524.878 baris dan tidak memiliki UnitPrice negatif. Kedua baris tetap disimpan untuk menjaga keterlacakan data.
